```python
# ============================================================
# CELL 13b: DIRECT SEQUENTIAL FL LOOP (bypasses Ray deadlock)
# ============================================================

import time
import logging
from pathlib import Path
from typing import Dict, Any, List

import flwr as fl
from flwr.common import FitRes, Status, Code, ndarrays_to_parameters, parameters_to_ndarrays

from evefl.fl.client import EveFLClient
from evefl.fl.dataset import get_hospital_dataloader
from evefl.fl.server import (
    build_initial_parameters, create_strategy, save_training_log,
    configure_logging, set_global_seed, get_device,
    DEFAULT_FRACTION_FIT, DEFAULT_MIN_FIT_CLIENTS, DEFAULT_MIN_AVAILABLE_CLIENTS
)


# --- Minimal proxy classes so the Strategy never touches Ray ---
class DirectProxy:
    def __init__(self, cid: str, client: EveFLClient):
        self.cid = cid
        self.client = client

class MockClientManager:
    def __init__(self, proxies: List[DirectProxy]):
        self._proxies = proxies
    def num_available(self) -> int:
        return len(self._proxies)
    def sample(self, num_clients: int, min_num_clients: int = 1) -> List[DirectProxy]:
        return self._proxies[:num_clients]


def run_experiment_direct(
    *,
    data_root: Path,
    partition_root: Path,
    num_clients: int,
    num_rounds: int,
    local_epochs: int,
    batch_size: int,
    n_qubits: int,
    intercept_probability: float,
    seed: int,
    experiment_name: str,
    output_path: Path,
) -> Dict[str, Any]:

    # --- validation ---
    if num_clients < 1: raise ValueError("num_clients must be >= 1.")
    if num_rounds < 1: raise ValueError("num_rounds must be >= 1.")
    if not 0.0 <= intercept_probability <= 1.0:
        raise ValueError("intercept_probability must be in [0,1].")

    set_global_seed(seed)
    configure_logging(verbose=False)
    log = logging.getLogger("evefl.server")

    log.info("=" * 60)
    log.info("EveFL DIRECT | %s", experiment_name)
    log.info("Device: %s | Clients: %d | Rounds: %d | Epochs: %d",
             get_device(), num_clients, num_rounds, local_epochs)
    log.info("Batch: %d | Qubits: %d | Intercept: %.2f", batch_size, n_qubits, intercept_probability)
    log.info("=" * 60)

    # --- build strategy ---
    initial_parameters = build_initial_parameters()
    strategy = create_strategy(
        initial_parameters=initial_parameters,
        intercept_probability=intercept_probability,
        n_qubits=n_qubits,
        fraction_fit=DEFAULT_FRACTION_FIT,
        min_fit_clients=min(num_clients, DEFAULT_MIN_FIT_CLIENTS),
        min_available_clients=max(num_clients, DEFAULT_MIN_AVAILABLE_CLIENTS),
    )

    # --- create clients directly (NO Ray, NO fork) ---
    log.info("Spawning %d clients directly...", num_clients)
    clients = []
    for i in range(num_clients):
        train_loader = get_hospital_dataloader(
            data_root=data_root,
            partition_root=partition_root,
            hospital_id=i,
            batch_size=batch_size,
            train=True,
        )
        client = EveFLClient(client_id=i, train_loader=train_loader)
        clients.append(client)
        log.info("  Client %d ready  (%d batches)", i, len(train_loader))

    proxies = [DirectProxy(str(i), clients[i]) for i in range(num_clients)]
    mock_manager = MockClientManager(proxies)

    # --- manual FL loop ---
    parameters = initial_parameters
    start_time = time.perf_counter()

    for round_num in range(1, num_rounds + 1):
        log.info("[ROUND %d] Starting...", round_num)

        # 1. Server decides policy + samples clients
        fit_instructions = strategy.configure_fit(round_num, parameters, mock_manager)

        # 2. Each client trains locally (sequential, same process)
        results = []
        failures = []
        for proxy, fit_ins in fit_instructions:
            # The original strategy hardcodes local_epochs=5; respect the arg we passed
            fit_ins.config["local_epochs"] = local_epochs

            try:
                params_ndarrays = parameters_to_ndarrays(fit_ins.parameters)
                updated_params, num_examples, metrics = proxy.client.fit(
                    params_ndarrays, fit_ins.config
                )

                fit_res = FitRes(
                    status=Status(code=Code.OK, message="Success"),
                    parameters=ndarrays_to_parameters(updated_params),
                    num_examples=num_examples,
                    metrics=metrics,
                )
                results.append((proxy, fit_res))
                log.info("[ROUND %d] Client %s done | loss=%.4f | examples=%d",
                         round_num, proxy.cid, metrics.get("loss", 0), num_examples)
            except Exception as exc:
                log.error("[ROUND %d] Client %s FAILED: %s", round_num, proxy.cid, exc)
                failures.append(exc)

        # 3. Server aggregates
        new_parameters, agg_metrics = strategy.aggregate_fit(round_num, results, failures)
        if new_parameters is not None:
            parameters = new_parameters

        state = agg_metrics.get("state", "UNKNOWN")
        qber  = agg_metrics.get("qber", 0.0)
        log.info("[ROUND %d] Aggregate complete | State=%s | QBER=%.4f", round_num, state, qber)

    elapsed_seconds = time.perf_counter() - start_time
    log.info("Finished in %.2f seconds.", elapsed_seconds)

    # --- export JSON (identical format to original) ---
    save_training_log(
        strategy=strategy,
        output_path=output_path,
        experiment_name=experiment_name,
        num_clients=num_clients,
        num_rounds=num_rounds,
        local_epochs=local_epochs,
        batch_size=batch_size,
        n_qubits=n_qubits,
        intercept_probability=intercept_probability,
        seed=seed,
        elapsed_seconds=elapsed_seconds,
    )

    secure   = sum(1 for r in strategy.round_logs if r.get("state") == "SECURE")
    caution  = sum(1 for r in strategy.round_logs if r.get("state") == "CAUTION")
    lockdown = sum(1 for r in strategy.round_logs if r.get("state") == "LOCKDOWN")

    summary = {
        "experiment": experiment_name,
        "elapsed_seconds": elapsed_seconds,
        "rounds_recorded": len(strategy.round_logs),
        "secure_rounds": secure,
        "caution_rounds": caution,
        "lockdown_rounds": lockdown,
        "output": str(output_path),
    }

    log.info("=" * 60)
    log.info("SECURE=%d  CAUTION=%d  LOCKDOWN=%d", secure, caution, lockdown)
    log.info("Results: %s", output_path)
    log.info("=" * 60)
    return summary


# ============================================================
# RUN THE FOUR SCENARIOS
# ============================================================

all_summaries = []

for name, intercept in SCENARIOS:
    print("\n" + "=" * 60)
    print(f"SCENARIO: {name}  |  intercept = {intercept}")
    print("=" * 60)

    summary = run_experiment_direct(
        data_root=KAGGLE_DATA,
        partition_root=PARTITION_ROOT,
        num_clients=NUM_CLIENTS,
        num_rounds=NUM_ROUNDS,
        local_epochs=LOCAL_EPOCHS,
        batch_size=BATCH_SIZE,
        n_qubits=N_QUBITS,
        intercept_probability=intercept,
        seed=42,
        experiment_name=name,
        output_path=RESULTS_DIR / f"{name}.json",
    )
    all_summaries.append(summary)
    print(f"\n✅ Saved: {summary['output']}")

print("\n" + "=" * 60)
print("ALL SCENARIOS COMPLETE")
print("=" * 60)
for s in all_summaries:
    print(f"{s['experiment']:12s} | {s['elapsed_seconds']:6.1f}s | "
          f"SECURE={s['secure_rounds']} CAUTION={s['caution_rounds']} LOCKDOWN={s['lockdown_rounds']}")
```
